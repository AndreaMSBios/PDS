# Variables and data types

Variables are used to save values for later. Imagine them as empty boxes filled with something. To "fill" a box, we use a name for our box, then `=` (the declaration sign, not the equality!) and then a value. This is called variable declaration or definition.

```python
x = 1
```

```python
x
```

Put it like this. If we empty the box, then we can put anything else in that box. In Python, the box is not explicitly emptied, we just put another value for the same variable name, and the "old" one will be forgotten.

```python
x = 2
```

The value of <code>x</code> has been changed.

```python
x
```

Notice that values in the previous examples were numerical (integers). What else value can be?

```python
x = "Hello world"
y = 1234
z = 3.14
k = True
```

## Basic data types

A variable has its type. Possible variable types in Python are:
- `str` (string) used for text or sequence of characters. For example, `"Hello world"` or `"1234"` (notice quotes)
- `int`, integers, e.g., `1234`
- `float`, used for real numbers (with dot not comma), e.g., `3.14` or `1.0`
- `bool`, used for logical values, e.g. `True` or `False`

When you declare or define variable, Python itself tries to infere variable data type. You don't have to do that explicitly. You can check variable type with command `type()` by putting variable name or value in the brackets.

If you need to convert variable type, for example, from a string to a number, use `int()` or `float()` on the string, depending on what you need; or conversely if you need to convert from a number to a string use `str()` on the number. Let's see it in the code:

```python
var = "1234"
type(var)
```

```python
var_int = int(var)
type(var_int)
```

Command `print()` can be used in combination with a value of any type mentioned above, explicitly:

```python
print('Hello world')
print(1234)
print(3.14)
print(True)
```

Or implicitly, by forwarding a variable name:

```python
print(x)
print(y)
print(z)
print(k)
```

## Format print and strings

For formatting the print (and strings), there are several options currently available in Python, but we will mention only two: by using `format()` command or by using special string literals *f-strings* (Python >= 3.6). We recommend the latter one, since it is much easier. Further, both will be presented.

### `format()`

The `format()` method formats the specified value(s) according to the provided format type.

```python
pi = 3.14159265
print(format(pi, '.2f'))
```

In the above example `pi` variable is formatted and print as fix point number (`.f`) with two decimals after the point (`'.2f'`), even if it has more than two decimal places after the point.

### Positional Formatting

`format()` can be used to insert variables inside the string's placeholder, too (not recommended). The placeholder is defined using curly brackets `{}`, in a dynamic string.

```python
txt = "I'm {age} years old.".format(age=49)
print(txt)
```

The placeholders can be identified using named indexes `{price}`, numbered indexes `{0}`, or even empty placeholders `{}`.

```python
txt1 = "My name is {name} and I'm {age} years old.".format(name='John', age=49)
txt2 = "My name is {0} and I'm {1} years old.".format('John', 49)
txt3 = "My name is {} and I'm {} years old.".format('John', 49)
print(txt1)
print(txt2)
print(txt3)
```

Inside the placeholders you can add a formatting type to format the result:

```python
price = 49.9878
print('The average price for our product is {:.2f}'.format(price))
```

## f-strings

This is the latest or new generation way of having dynamic string. Use `f` before string instead of using `.format()` after it. *f-string* is similar to the named formatting (the keyword to the actual variable name as a placeholder).

```python
name = 'John'
age = 49
txt = f"My name is {name} and I'm {age} years old."
print(txt)
```

```python
price = 49.9878
print(f'The average price for our product is {price:.2f}')
```
