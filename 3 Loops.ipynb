## `for` loop

Very useful Python function for iterating numbers (integers) is `range()` function. See what it is and what are the accepted arguments (values inside the brackets):

```python
range?
```

<div class="alert alert-block alert-warning">
<b>Example: </b> Suppose we want Python to print the numbers from 0 to 4. Instead of writing five separate <code>print()</code> statements, we can use a <code>for</code> loop to repeat the same action for each number.
</div>

```python
for i in range(5):    #  list(range(5)) returns list [0, 1, 2, 3, 4]
    print(i)
```

<div class="alert alert-block alert-warning">
<b>Example: </b> Create a list and iterate over it.
</div>

```python
primes = [2, 3, 5, 7]   # creates new list
for prime in primes:
    print(prime)
```

<div class="alert alert-block alert-warning">
<b>Example: </b> Print each character from hello_world.
</div>

```python
hello_world = "Hello, World!"
for ch in hello_world:
    print(ch) # space is character
```

Check string length and compare with number of printed characters

```python
len(hello_world)
```

<div class="alert alert-block alert-info">
<b>Tip: </b>You can use <code>len()</code> to check length for both strings and lists.
</div>

<div class="alert alert-block alert-warning">
    <b>Example: </b>Calculate string length using <code>for</code> loop.
</div>

```python
length = 0      # initializes length variable
for ch in hello_world:
    length += 1     # adds 1 to the length on each iteration (or length = length + 1)

print(len(hello_world), length)
```

### Iterating by Sequence Index

<div class="alert alert-block alert-warning">
<b>Example: </b>Iterate through list by its index. Print each item.
</div>

```python
fruits = ['banana', 'apple', 'mango']
for index in range(len(fruits)):
    print('Current fruit :', fruits[index])
```

### Nested `for` loop

You can use one `for` loop inside the other one. This is called nested `for` loop. While fixing first value for the outside `for` loop, we iterate through all values in the inner one, and only if we pass all values in the inner, we can switch to the next value in the outside loop. The procedure is repeated until we reach the end of the iterable (e.g., `list`, `range`) in the outside loop.

<div class="alert alert-block alert-warning">
<b>Example: </b>Create a multiplication chart of 10 i.e., print numbers in a form of a 10 x 10 table, where each value is a number from 1 - 10 (column-wise) multiplied with number from 1 - 10 (row-wise).
</div>

```python
for i in range(1,11):
    for j in range(1,11):
        k = i * j
        print(k, end='\t') # \n
    print() # used for new line printing
```

### List comprehensions

List comprehensions are very Pythonic way for creating lists using `for` loop.

<div class="alert alert-block alert-warning">
    <b>Example: </b>Create a list containing the squares of all numbers less than 10.
</div>

There are two ways to achieve this. A bit faster is the solution with list comprehension.

```python
squares = []
for x in range(10):
    squares.append(x**2)
print(squares)
```

OR

```python
squares = [x**2 for x in range(10)]  # use list comprehensions to get the same result
print(squares)
```

<div class="alert alert-block alert-warning">
    <b>Example: </b>Create a list of numbers under 50 divisible by 2.
</div>

```python
evens_to_50 = [i for i in range(51) if i % 2 == 0]
print(evens_to_50)
```

<div class="alert alert-block alert-warning">
<b>Example: </b>Create a list of lists containing the products of numbers from 1 to 3.
</div>

```python
table = []

for i in range(1, 4):
    row = []
    for j in range(1, 4):
        row.append(i * j)
    table.append(row)

print(table)
```

OR

```python
table = [[i * j for j in range(1, 4)] for i in range(1, 4)]
print(table)
```

<div class="alert alert-block alert-warning">
<b>Example: </b>Given a dictionary containing students and their exam scores, sort the students by score in descending order and create a list containing the names of students who scored 80 or more.
</div>

```python
scores = {"Ana": 85, "Marko": 72, "Sara": 95, "Luka": 68}
```

```python
sorted_scores = sorted(scores.items(), key=lambda x: x[1], reverse=True)
print(sorted_scores)
```

```python
top_students = [name for name, score in scores.items() if score >= 80]
print(top_students)
```

## `while` loop

<code>for</code> loop: repeat for each item or a known sequence of iterations.

<code>while</code> loop: repeat while a condition remains true.

<div class="alert alert-block alert-warning">
<b>Example: </b>Keep asking the user to enter a password until the correct password is provided.
</div>

```python
password = ""

while password != "python123":
    password = input("Enter the password: ")

print("Access granted!")
```

<div class="alert alert-block alert-warning">
<b>Example: </b>Print all squares from 0 to 99.
</div>

```python
square = 0
number = 1
while square <= 99:  #
    square = number ** 2
    print(square)
    number += 1
```

<div class="alert alert-block alert-warning">
<b>Example: </b>Allow the user to enter numbers until they enter "q", then display the numbers.
</div>

```python
numbers = []

while True:
    value = input("Enter a number or 'q' to quit: ")

    if value == "q":
        break

    numbers.append(float(value))

print("Numbers:", numbers)
```
