# Functions

You can use functions in programming to store a piece of code that can be invoked when needed. This prevents you from retyping the same logic every time you need that code.

<div class="alert alert-block alert-warning">
<b>Example: </b>Create function named "hello_world" that prints "Hello, World!".
</div>

```python
def hello_world(x):  
    print("Hello, World!")
```

<code>def</code> is a Python keyword used to define a function, followed by the function's name. Notice that the function name is highlighted in blue, even though it is not a Python keyword or reserved word. The name is simply recognized as a function name after it is defined.

After the function name, its arguments (or parameters) are specified in parentheses. Arguments allow us to pass information to the function. A function can have as many arguments as needed; they are separated by commas.

<div class="alert alert-block alert-warning">
<b>Example: </b>Use the above created function 5 times.
</div>

```python
for i in range(5):
    hello_world(i)
```

In the example above, only <code>x</code>  is passed to the function, but it is not used inside the function.

<div class="alert alert-block alert-warning">
    <b>Example: </b>Pass a number to the function that prints that number next to text "x = ".
</div>

```python
def foo(x):                 # x is a function parameter
    print("x = " + str(x))
```

```python
foo(5)   # pass 5 to foo(). Here 5 is an argument passed to function foo.
```

It is recommended to write a notes for a function that explains what it is used for, what arguments it accepts and/or what it returns. The next example shows .

```python
def fib(n):
    """This is documentation string for function. It'll be available by fib.__doc__()
    Return a list containing the Fibonacci series up to n."""
    
    result = []
    a = 0
    b = 1
    while a < n:
        result.append(a)
        tmp_var = b
        b += a
        a = tmp_var
    return result
```

```python
fib(20)
```

# Lambda functions

An anonymous function in Python is a function without a name. It can be immediately invoked or stored in a variable. Anonymous functions in Python are also known as **lambda** functions.

The syntax for its creation is `lambda parameter(s): expression`

- A lambda function is created using the lambda keyword.
- The keyword is followed by one or many parameters.
- Lastly, an expression is provided for the function. This is the part of the code that gets executed/returned.

<div class="alert alert-block alert-warning">
<b>Example: </b>Create an anonymus function that adds two numbers
</div>

```python
add_numbers = lambda a,b : a + b 
print(add_numbers(2,3))
```

# Type hints

In the example for the function that adds two numbers:

```python
def sum_two_numbers(a, b):
    return a + b  
```

We can pass both numbers or strings to this function, without any error.

```python
sum_two_numbers(2,3)
```

```python
sum_two_numbers('Hello','world')
```

But what if we want to limit the function to accept only numbers? And restrict only on floats, for example.

Recently, type hints were added to the language and now you can write syntactically correct function specifications including the types of arguments and return values, even dough it is not mandatory.

<div class="alert alert-block alert-warning">
<b>Example: </b>The function below takes and returns a string. It is annotated with types.
</div>

```python
def greeting(name: str) -> str:
    return 'Hello ' + name
```

In the function `greeting`, the argument `name` is expected to be of type `str` and the return type `str`.
