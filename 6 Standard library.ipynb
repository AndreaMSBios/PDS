# Random module

```python
import random
```

Useful functions in this module:
- `random.random()` - returns float random number less than 1 and greater or equal to 0
- `random.uniform(a, b)` generates a floating point random number between the numbers [a,b)
- `random.gauss(mu, sigma)` - generates random number from Gaussian distribution; mu is the mean, and sigma is the standard deviation
- `random.randint(a, b)` - returns a random integer N such that a <= N <= b
- `random.randrange(beg, end, step)` - generates random number, takes three arguments: beginning number (included in generation), last number (excluded in generation) and step (to skip numbers in range while selecting)

<div class="alert alert-block alert-warning">
<b>Example: </b>Generate 5 numbers on random.
</div>

```python
for i in range(5):
    print(random.random()) # random() 
```

Other useful functions on a sequence:
- `random.shuffle(list)` - shuffle the entire list to randomly arrange its elements
- `random.choice(list)` - generates a random element from a given list
- `random.sample(population, k)` - returns a k length list of unique elements chosen from the population sequence or set. Used for random sampling without replacement

Random numbers on a computer are not as random as they seem - they are pseudorandom. A random seed (or seed state, or just seed) is a number (or vector) used to initialize a pseudorandom number generator. It's just a one way to "capture" the random value. Random seeds are often generated from the state of the computer system (such as the time). It is not important what the value is set as input to the `random.seed()` function, but to use it every time when we want to get the same result from Python random module. 

<div class="alert alert-block alert-warning">
<b>Example: </b>Set seeds before generating two random numbers. Repeat one number as input to the <code>random.seed()</code> two times to get the same random number.
</div>

```python
random.seed(5) # Initialize the random number generator
print('Random number with seed = 5:', random.random())
random.seed(7) 
print('Random number with seed = 7:', random.random())
random.seed(5) 
print('Random number with seed = 5:', random.random())
```

# Datetime module

For manipulating with date and time in Python, `datetime` module is used.

```python
import datetime
```

## Date and time objects

To see the current date and time use function `now()` from `datetime` module and its `datetime` type. Use:

```python
datetime.datetime.now()
```

As you can notice, the given output is a datetime object with a series of numbers, but if you are unsure what they represent of you can always check in the documentation. To get better human-eye output format, use `print`.

```python
d = datetime.datetime.now()
print(d)
```

Now, you can guess the year, month, day, hour... Since variable `d` is a datetime object, it has these attributes as accessible fields: 

```python
d.year, d.month, d.day, d.hour, d.microsecond, d.tzinfo 
```

A `datetime` object has a method for formatting date objects into readable strings. The method is `strftime()`. It takes one parameter `format` to specify the format of the returned string. To specify `format`, format codes are used. For example, use code `%y` for abbreviated year, a year without century as a zero-padded decimal number:

```python
year = d.strftime("%y")
year
```

<div class="alert alert-block alert-warning">
<b>Example: </b>Specify date of the following format: 
    <ol>
        <li>Day is a zero-padded decimal, month is a word and year is given as a full year; </li>
        <li>They are separated with a slash <code>/</code>.</li>
    </ol>
</div>

```python
x = datetime.datetime(2023, 5, 24).strftime('%d/%B/%Y')
x
```

Likewise, there is a method `strptime()` for creating a `datetime` object from the given string. It takes date string and format code as input parameters, and returns a `datetime` object.

## `timedelta` object type

A `timedelta` object represents a duration, the difference between two dates or times.

<div class="alert alert-block alert-warning">
<b>Example: </b>Define one day difference.
    <ol>
        <li>Add it and subtract from today, to get a yesterday and tomorrow date. </li>
        <li>Subtract tomorrow and yesterday, and yesterday and tomorrow dates to see the output.
    </ol>
</div>

```python
print('Today  :', today)
one_day = datetime.timedelta(days=1)
print('One day  :', one_day)

yesterday = today - one_day
print('Yesterday:', yesterday)

tomorrow = today + one_day
print('Tomorrow :', tomorrow)

print('tomorrow - yesterday:', tomorrow - yesterday) 
print('yesterday - tomorrow:', yesterday - tomorrow) 
```

Dates can be compared:

```python
yesterday > today, tomorrow > today # Comparing Values
```
