## Assignment and Augmented Assignment Operators

Assignment operators are used to assign values to variables.

```python
x = 5
```

Update a variable by performing an operation on its current value. Write it as a shorthand.

```python
x += 2 # same as x = x + 2
x
```

You can use other operators in combination with `=`. These are shorthand notations for updating the left operand in place: `-=`, `*=`, `/=`, `//=` and `%=`

## Membership Operators

Operator `in` checks if something is in some object (string, list, dictionary etc.). Returns `True` if a sequence with the specified value is present in the object.

```python
grocery_string = 'We have fish, tomato and apples.'
```

```python
'tomato' in grocery_string
```

But be careful Python is case-sensitive

```python
'We' in grocery_string
```

```python
'we' in grocery_string
```

## Logical (boolean) operators: `and`, `or`, `not`

Logical operators are used to combine conditional statements. `and` and `or` expect two operands to work with, typically referred to as the left operand and the right operand, while `not` is unary operator.

```python
name = 'John'
age = 17
```

<div class="alert alert-block alert-warning">
<b>Example: </b> Check if name is John and its age is not more than 17
</div>

```python
name == 'John' or not age > 17
```

You can assign the above conditional statement to the variable to be used further.

```python
g = (name == 'John' and age > 17)
```

<div class="alert alert-block alert-info">
<b>Note: </b>When you use the boolean <code>and</code> operator the second expression is not evaluated when the first is <code>False</code>. Similarly <code>or</code> does not evaluate the second argument if the first is <code>True</code>.
</div>

<div class="alert alert-block alert-info">
<b>Note: </b>You can use bitwise operators <code>&</code>, <code>|</code>, <code>~</code> to perform Boolean logic on individual bits. That’s analogous to using logical operators <code>and</code>, <code>or</code> and <code>not</code>, but on a bit level. The similarities between bitwise and logical operators go beyond that. When you use the (binary) bitwise operators both expressions are evaluated.
</div>

# Conditions

Conditions are written in the form:
<code>
if condition:
    statement 
elif condition:
    statement 
else
    statement 
</code>

<div class="alert alert-block alert-warning">
<b>Example: </b> Imagine we want to check whether a person is old enough to enter an event. If their age is 18 or more, they can enter.
</div>

```python
age = 20

if age >= 18:
    print("You can enter.")
```

You can add an alternative.

```python
age = 20

if age >= 18:
    print("You can enter.")
else:
    print("You cannot enter.")
```
