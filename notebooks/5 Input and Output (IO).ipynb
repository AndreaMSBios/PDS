## Reading files

Use function `open` to read files. It returns a text stream (a file object or a file-like object) with mode string 'r' and byte stream (also called buffered I/O) with mode string 'rb'. Text streams demand encoding and decoding, as well as translation of platform-specific [newline characters](https://jdhao.github.io/2018/12/08/newline_vim_python_sublime_notepad/), while byte streams not. There is also one more stream, which is rarely used (raw stream, also called unbuffered I/O).

```python
open?
```

<div class="alert alert-block alert-warning">
    <b>Example: </b>Read <em>input.txt</em> file.
</div>

```python
f = open('input.txt', 'r')   # here we open file "input.txt". Second argument used to identify that we want to read file
for line in f.readlines():   # read lines
    print(line)
f.close()                    # It's important to close the file to free up any system resources.
```

<div class="alert alert-block alert-danger">
<b>Be careful: </b> Don't mix <code>f.readlines()</code> and <code>f.readline()</code>!
</div>

<div class="alert alert-block alert-danger">
<b>Be careful: </b> Don't forget to close the already opened file! This will be more important when using 'w' mode for writing files.
</div>

If you don't want to think about closing the file, use `with` statement:

```python
with open('input.txt', 'r') as f:  # the file is properly closed after its suite finishes
    for line in f.readlines():
        print(line)
```

## Pickle module

`pickle` - binary protocols for serializing and de-serializing a Python object structure; simply, used for saving the big data (See [this](https://docs.python.org/3/library/pickle.html) for more information)

<div class="alert alert-block alert-warning">
<b>Example: </b>Write a pickled file.
</div>

```python
a = ['test 1','test 2','test 3']
f = open('my_file.pkl','wb')
pickle.dump(a,f) # The pickle module takes a Python data structure and saves it to a .pkl file
f.close()
```

<div class="alert alert-block alert-warning">
<b>Example: </b>Read the pickled file.
</div>

```python
f = open('my_file.pkl','rb')
b = pickle.load(f) # reads the serialized data
f.close()
```

## Json module

JSON (JavaScript Object Notation) is a text format that is completely language independent but uses conventions that are familiar to programmers of the C-family of languages, including C, C++, C#, Java, JavaScript, Perl, Python, and many others. These properties make JSON an ideal data-interchange language.

JSON is built on two structures:
- A collection of name/value pairs. In various languages, this is realized as an object, record, struct, dictionary, hash table, keyed list, or associative array.
- An ordered list of values. In most languages, this is realized as an array, vector, list, or sequence.

An example of JSON string would be:

'{"name":"John", "age":30, "car":null}'

or

[John, Doe]

(See [this](https://www.json.org/json-en.html) for more information)

`json` - a Python library for JSON manipulation

JSON files are human-readable, while pickle are not

(See [this](https://docs.python.org/3/library/os.html) for more information)
