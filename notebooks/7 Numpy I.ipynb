# Creating arrays

The NumPy ndarray class is used to represent both matrices and vectors. A vector is an array with a single dimension (there's no difference between row and column vectors), while a matrix refers to an array with two dimensions. For 3-D or higher dimensional arrays, the term tensor is also commonly used.

<div class="alert alert-block alert-warning">
<b>Example: </b>Create a NumPy array containing the exam scores of five students. Check the array's dimensions, shape, size, and data type. Calculate the average score.
</div>

```python
scores = np.array([78, 85, 92, 67, 88])

print("Array:", scores)
print("Number of dimensions:", scores.ndim)
print("Shape:", scores.shape)
print("Number of elements:", scores.size)
print("Data type:", scores.dtype)

print("Average score:", np.mean(scores))
```

## Data types

All of the elements in a NumPy array should be homogeneous. The type of its elements is defined through `dtype` attribute. By default, NumPy tries to infer `dtype` from an array, but it can be specified while defining. Use convertible strings or `dtype` objects for specifying the type of array elements.

```python
a = np.array([0, 1, 2, 3])
b = np.array([0, 1, 2, 3], dtype='float64')
a, b
```

```python
c = np.array(['Hello', 'world', '!'])
c.dtype # see inferred convertible string for array of strings
```

For more information on dtype and convertible strings use the __[dtype documentation](https://numpy.org/doc/stable/reference/arrays.dtypes.html)__.

The `numpy.meshgrid()` function is used to create a rectangular grid out of two given one-dimensional arrays representing the Cartesian indexing or Matrix indexing.

<img src="https://i.stack.imgur.com/8Mbig.png" alt="A meshgrid" title="A meshgrid" />

<div class="alert alert-block alert-warning">
<b>Example: </b>Suppose a shop sells products at different prices, and we want to calculate the total cost for different quantities.
</div>

```python
quantities = np.array([1, 2, 3, 4])
prices = np.array([10, 20, 30])

Q, P = np.meshgrid(quantities, prices)

total = Q * P
```

<code>np.meshgrid()</code> creates a grid containing every possible combination of quantities and prices.

rows = different prices

columns = different quantities

each cell = price × quantity

The resulting <code>total</code> array shows the cost for each combination

```python
print(total)
```

# Indexing and slicing NumPy arrays

For specifying one or multiple values from an NumPy array, we use indexing and slicing. Indexing and slicing on 1-D arrays are similar to indexing and slicing Python lists (with square-brackets). Lets define an arbitrary array:

<div class="alert alert-block alert-warning">
<b>Example: </b>Create a NumPy array containing the numbers from 0 to 10. Access the fifth element, select the elements with indices from 5 to 8, and replace all values in that range with 1.
</div>

```python
data = np.arange(11)

print(data[4]) # Show the fifth element
print(data[5:9])# Show elements with indices 5 to 8 (slice it)
```

```python
data[5:9] = 1# Replace elements with indices 5 to 8
print(data)
```

<div class="alert alert-block alert-danger">
<b></b>Array slices are views on the original array! Data is not copied, and any modifications to the view will be reflected in the source array.
</div>


```python
data_slice = data[5:9]  # Save the slice as variable
data_slice[1] = 12345 # Change second element in the slice
data
```

For indexing and slicing n-dimensional arrays, you can use either multiple square brackets or comma-separated indices within a single pair of square brackets. When an index or slice is specified for one axis, NumPy returns all values along the other axes unless they are also indexed or sliced. See the example below.

<div class="alert alert-block alert-warning">
<b>Example: </b>Consider a 2D NumPy array representing measurements from three samples at four time points:
    
1. Select the second row
2. Select the third column9
3. Select a specific element (e.g., element in the second row and the third column)
4. Slice several rows and columns (e.g., first two rows and last three rows)
</div>


```python
data = np.array([
    [10, 12, 15, 18],
    [20, 22, 25, 28],
    [30, 32, 35, 38]])

print(data)
```

```python
print(data[1]) # OR print(data[1, :])
print(data[:, 2])
print(data[1, 2])
print(data[0:2, -3:])
```

<div class="alert alert-block alert-warning">
<b>Example: </b>Choose even and odd indexed elements from an array.
</div>

```python
arr = np.arange(20)
print('Even elements: ', arr[::2])
print('Odd elements: ', arr[1::2])
```

# Functions

There are some special functions in NumPy, performed element-wise on the passed array.

<div class="alert alert-block alert-warning">
<b>Example: </b>Create a 2D grid for x and y ranging from -5 to 5 with a step of 0.1. Use <code>np.meshgrid()</code> to calculate:
    $$ z=\sin(x^2+y^2) $$
</div>

```python
#solution
```

```python
x = np.arange(-5, 5.1, 0.1)
y = np.arange(-5, 5.1, 0.1)
xx, yy = np.meshgrid(x, y)
z = np.sin(xx**2 + yy**2)
print('Result ', z)
```
