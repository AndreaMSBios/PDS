# Comparison

Comparison is performed element-wise with standard comparison operators `==`, `>`, `<` etc.

<div class="alert alert-block alert-warning">
<b>Example: </b>Suppose three experiments produced measurements for the four samples. We want to check which values are equal between the experiments, if elements from the first sample are greater compared to the second one, and if the elements of the first two samples are different.
</div>

```python
experiment_A = np.array([10, 15, 20, 25])
experiment_B = np.array([10, 18, 20, 30])
experiment_C = np.array([10, 15, 20, 25])

print("A == B:", experiment_A == experiment_B)
print("A == C:", experiment_A == experiment_C)

print("A > B:", experiment_A > experiment_B)
print("A != B:", experiment_A != experiment_B)
```

Previous code gives True/False answer for each pair of elements at the specific position (element-wise). For one unique answer, use <code>np.array_equal()</code>. It checks if two arrays have the same shape and elements.

```python
np.array_equal(experiment_A, experiment_C)
```

You can then use <code>np.all()</code> to ask whether the entire arrays are identical.

```python
print(np.all(experiment_A == experiment_B))
print(np.all(experiment_A == experiment_C))
```

And <code>np.any()</code> can answer whether at least one value is equal.

```python
print(np.any(experiment_A == experiment_B))
```

<div class="alert alert-block alert-warning">
<b>Example: </b>(Multiple comparisons) Suppose we have the exam scores of several students and want to check if everyone passed with score between 50 and 90.
</div>

```python
scores = np.array([45, 62, 78, 91, 55, 88])
result = (scores > 50) & (scores < 90)
np.all(result)
```

<div class="alert alert-block alert-info">
    <b>Tip: </b>Logical operators (e.g., <code>and</code>, <code>or</code>, <code>!</code>) evaluate boolean conditions (True or False) and are typically used to control program flow in if statements. Bitwise operators (e.g., <code>&</code>, <code>|</code>, <code>~</code>) work on integers at the binary level, manipulating individual bits (1s and 0s). Look <a href="https://realpython.com/python-bitwise-operators/">here</a> for more information.

# Conditioning

Conditioning allows us to select only the elements of a NumPy array that satisfy a given condition. It is performed similar to slicing.

```python
temperatures = np.array([15, 17, 23, 26, 20])
```

```python
temperatures[temperatures > 23]  # Returns only the temperature values greater than 23.
```

Set the element-wise condition on NumPy array with optional choice of substitute value if condition is not met with <code>np.where()</code>.

<div class="alert alert-block alert-warning">
    <b>Example: </b>Suppose we have exam scores and want to replace all failing scores with 0.
</div>

```python
scores = np.array([45, 62, 78, 91, 55, 88])
result = np.where(scores < 50, 0, scores)
print(result)
```

Note that the arguments of <code>np.where()</code> can be arrays or specific values (such as numbers or strings) to select depending on whether the condition is met.

```python
result = np.where(scores < 50, 'Fail', scores)
print(result)
```

To find the corresponding indices of the given array if defined condition is met, use `np.argwhere()` function.

```python
np.argwhere(scores < 50).reshape(-1)
```

Since `np.argwhere()` returns 2-d array, we reshaped it to get the 1-d array.

# Basic reductions

Here we consider all NumPy functions which reduce an array into a single value, like `np.max()`, `np.min()`, `np.sum()`, `np.std()`, `np.mean()` and `np.median()`.

Take next array as an example array:

```python
temperatures = np.array([15, 17, 23, 26, 20, 5, 8, 27, 12, 21, 15])
```

Show maximum temperature measured:

```python
np.max(temperatures)
```

But you can also use the above mentioned functions directly on the array:

```python
temperatures.max()
```

If you are interested in the position of the minimum or maximum value use `np.argmin()` or `np.argmax()`:

```python
temperatures.argmin(), temperatures.argmax()
```

The situation changes a little bit with 2-D arrays. Let's take the next array as an example:

```python
scores = np.array([
    [18, 22, 25, 21, 19, 5],
    [15, 17, 23, 26, 20, 21],
    [20, 24, 27, 25, 22, 9]])
```

By using `np.sum()` on the 2-D array, total sum would be returned:

```python
scores.sum() # sums over all the dimensions of the input 
```

However, if we want sum by rows or sum by columns, we need to use parameter `axis`. This is a parameter along which a sum is performed.

```python
scores.sum(axis=0)  # Sum operates along rows (sum by columns)
```

```python
scores.sum(axis=1) # Sum operates along columns (sum by rows)
```

`axis` parameter can be used with all reduction functions.

# Sorting NumPy arrays

For sorting a NumPy array, you can use either `np.sort()` or the array method `.sort()`. If you need the indices that would sort the array, use `np.argsort()` or the array method `.argsort()`. These functions can be called either from NumPy or directly on the array.

```python
s1 = np.array([4, 3, 1, 2])
print('s1 = ', s1)
```

```python
np.argsort(s1), s1.argsort()
```

However, the default `axis` parameter is -1 (the last axis), which means that in the 2-D array case, that parameter would be `axis=1`. If you need to sort flattened array use `axis=None`.

<div class="alert alert-block alert-warning">
<b>Example: </b>Compare sorting along different values of the <code>axis</code> parameter.
</div>

```python
s2 = np.array([[4, 3, 5], [1, 2, 1]])
print('s2=', s2)
```

```python
np.sort(s2, axis=0) # Sorting along axis=0/rows
```

```python
np.sort(s2, axis=1) # Sorting along axis=1/columns
```

```python
s2.argsort() # Sorting along an axis=1 (default: last axis)
```

```python
s2.argsort(axis=None) # Sorting the flattened array
```

# Masked arrays

NumPy arrays can be sliced with other arrays (but not any!). Here we will refer to the such arrays as *masks*. For mask we can use integer-based arrays (or lists) of any shape or Boolean arrays of the same shape as obtained array.

In the next example, a list is used to select specific elements of the array by its index.

```python
slicer = [2, 4, 8, 15]
arr = np.random.rand(20)

print('Array: ', arr)
print('Sliced array: ', arr[slicer])
```

```python
arr = np.arange(20)
slicer = np.where(np.random.rand(arr.shape[0]) > 0.5, True, False)
print('Sliced array: ', arr[slicer])
```
