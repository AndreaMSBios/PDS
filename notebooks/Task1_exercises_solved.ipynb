# Assignment 1

Set all lowercase letters and remove punctation from a string. For example: 

Input:  ’ - Lorem; ipsum; Lorem;_ipsum?’

Output: ‘lorem ipsum loremipsum’

```python
s = '- Lorem; ipsum; Lorem;_ipsum?'
punct = ".,:;!?-_/\|><()[]{}*'"
s = ''.join(['' if ch in punct else ch.lower() for ch in s])
s = s.strip()
s
```

# Assignment 2

For some plumber job ideal pipe diameter and tolerance are determined by building standards. Write Python function which takes ideal diameter, allowed tolerance and pipe diameter for pipe delivered to the building company, and checks if it fits to the standards. 

```python
def diameter_check(ideal, tol, real):
    if abs(ideal - real) < tol:
        print('It fits to the standards')
    else:
        print("It doesn't fit to the standards")
```

```python
diameter_check(10, 1, 10.5)
```

# Assignment 3

Monopoly is a game with a round game board in which players move clockwise field-by-field (multiple circles are possible). Assume that fields are enumerated with numbers starting from 0, that this is an initial field for both players, and that there is no moving backward. If we know total number of fields, number of fields passed from the beginning by the first player and number of fields passed from the beginning by the second player, write a Python function for calculating number of steps necessary for the first player to reach the second one.

```python
def first_reach_second(total_no, first_no, second_no):
    position1 = first_no % total_no
    position2  = second_no % total_no
    steps_to_reach = position2 - position1
    
    if steps_to_reach < 0:
        steps_to_reach += total_no
        
    return steps_to_reach
        
```

```python
first_reach_second(20, 30, 22)
```

```python
f = open('history.txt', 'r')   # here we open file "input.txt". Second argument used to identify that we want to read file
d = {}
for line in f.readlines():   # read lines
    line = line.upper()
    for j in [',','.','!','?',')','(','1','2','3','4','5','6','7','8','9','0']:
        line = line.replace(j,' ')  
    for j in line.split():
        if len(j)>1:
            if j in d.keys():
                d[j] = d[j] + 1
            else:
                d[j] = 1
f.close()              
```

```python
def frequent_words(n):
    items = [(v, k) for k, v in d.items()] 
    items = sorted(items, reverse=True)
    items = [(k, v) for v, k in items]
    print(items[:n])
```

```python
frequent_words(5)
```
